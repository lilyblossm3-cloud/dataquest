```python
import pandas as pd

url="https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df=pd.read_csv(url)
print(f"Dataset Loaded Succesfully! Total rows and columns:{df.shape}")

df.head()
```

```python
print("--- DATA SUMMARY & TYPES ---")
df.info()
```

```python
print("....Missing values per column...")
print(df.isnull().sum())
print("..Missing values with percentage..")
(df.isnull().sum()/ len(df)*100).round((1))
```

```python
print("..Percentage of Suvived People..")
df["Survived"].value_counts()
df["Survived"].value_counts(normalize=True).round(3)
```

```python
print("..Percentage of survived based on Sex..")
df.groupby("Sex")["Survived"].mean()
```

```python
print("Based on Class")
df.groupby("Pclass")["Survived"].mean()
```

```python
numeric_cols=df.select_dtypes(include="number").columns.tolist()
categorical_cols=df.select_dtypes(include="object").columns.tolist()
print("Numeric:", numeric_cols)
print("Categorical:", categorical_cols)
```

```python
print("Correlation with numeric values")
df[numeric_cols].corr()["Survived"].sort_values(ascending=False)
```

```python
for col in categorical_cols:
  if df[col].nunique()<10:
    print(f"\n-- {col} ---")
    print(df.groupby(col)["Survived"].mean().sort_values(ascending=False))
```
